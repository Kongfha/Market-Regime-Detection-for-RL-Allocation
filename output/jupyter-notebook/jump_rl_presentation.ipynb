# Jump Model + RL Tuning Presentation Figures

This notebook collects presentation-ready charts for the Jump Model regime detector and the planned RL tuning workflow.

Key headline from the current validation-only baseline check: `momentum_rotation_20d` has validation Sharpe `0.3938` and cumulative return `16.9%`. The tuned RL sweep has not been run yet; its scores will appear in `output/jump_rl_tuning/` after training.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
weekly = pd.read_csv(ROOT / 'data/processed/jump_model_train_ready_weekly.csv', parse_dates=['week_end'])
validation_scores = pd.read_csv(ROOT / 'output/jump_rl_presentation/baseline_validation.csv')
locked_scores = pd.read_csv(ROOT / 'output/jump_rl_presentation/baseline_locked_test.csv')
weekly.head()
```

## 1. Regime Timeline

Use this chart to explain what the Jump Model is doing before RL enters the story: it converts weekly market features into interpretable market states, then those states become part of the RL observation.

![Regime timeline](../jump_rl_presentation/regime_timeline.svg)

## 2. Regime Coverage By Train / Validation / Test

This is a useful credibility chart: each split contains multiple regimes, so validation and locked test are not just one market condition.

![Regime split coverage](../jump_rl_presentation/regime_split_coverage.svg)

```python
summary = (
    weekly.groupby(['split', 'regime_name'])
    .size()
    .rename('weeks')
    .reset_index()
    .sort_values(['split', 'regime_name'])
)
summary
```

## 3. Validation Baseline Scores

This is the scorecard before RL tuning. The selected RL model should beat these on validation Sharpe, not on locked-test hindsight.

![Validation baseline scores](../jump_rl_presentation/validation_baseline_scores.svg)

```python
validation_scores.sort_values('sharpe_ratio', ascending=False)
```

## 4. Locked-Test Baseline Context

This chart is for final context only. It should not be used to select the RL model. Current top locked-test baseline by Sharpe is `gld_only` with Sharpe `1.7702`.

![Locked test baseline scores](../jump_rl_presentation/locked_test_baseline_scores.svg)

```python
locked_scores.sort_values('sharpe_ratio', ascending=False)
```

## 5. RL Tuning Funnel

This is the slide that explains the experimental protocol: broad screen, longer rerun, multi-seed finalist check, then one locked-test report.

## Suggested Presentation Flow

1. Start with the regime timeline: the model identifies persistent market states.
2. Show split coverage: the experiment has regime variety across train, validation, and test.
3. Show validation baselines: this is the bar RL must clear.
4. Show the tuning funnel: model selection is validation-only.
5. End with locked-test context after the sweep runs.
